```python
#digit count
q=int(input("Enter  a number:"))
sum=0
while(q!=0):
  rem=q%10
  sum=sum+rem
  q=q//10
print("sum of digit are",sum)
```

```python
#digit count with function
def sod(q):
  sum=0
  while(q!=0):
    rem=q%10
    sum=sum+rem
    q=q//10
  print("sum of digit are",sum)

q=int(input("Enter  a number:"))
sod(q)
```
